```python
vcols = [c for c in train_df.columns if c.startswith("VelTrace_")]
tcols = [c for c in train_df.columns if c.startswith("VelTime_")]
first_v_index = train_df.columns.get_loc(vcols[0])
last_v_index = train_df.columns.get_loc(vcols[-1])
first_t_index = train_df.columns.get_loc(tcols[0])
last_t_index = train_df.columns.get_loc(tcols[-1])
rt3_index = train_df.columns.get_loc("rt3")
```

```python
all_v_probs = []

for i in range(len(train_df)):    
    t = train_df.iloc[i, first_t_index:last_t_index].to_numpy(dtype=float)
    rt3 = train_df.iloc[i, rt3_index]
    if rt3 > 0:
        v_prob = (t >= rt3).astype(np.float32)
    else:
        v_prob = np.zeros_like(t, dtype=np.float32)
    all_v_probs.append(v_prob)
        
all_v_probs = np.array(all_v_probs)    
np.save("../data/train_prob_values.npy", all_v_probs)
```
